```python
import sqlite3
```

```python
conn = sqlite3.connect("supply_chain.db")
```

```python
df.to_sql("orders", conn, if_exists='append')
```

```python
query = "SELECT * FROM orders;"
pd.read_sql_query(query, conn)
```

```python
query = 'SELECT "Order ID", Sales FROM orders;'
pd.read_sql_query(query,conn)
```

```python
query = 'SELECT "Order ID", Sales FROM orders WHERE Sales > 1000;'
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales, Market FROM orders WHERE Market = 'EUROPE';'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales FROM orders ORDER BY Sales DESC;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales FROM orders ORDER BY sales DESC LIMIT 10;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT Market, SUM(Sales) AS "Total Sales" FROM orders GROUP BY Market;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Market", SUM(Sales) AS Total_Sales FROM orders 
GROUP BY Market
HAVING SUM(Sales) > 10000000;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Market", AVG(Sales) as Average_Sales FROM orders
GROUP BY Market;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT Market, COUNT("Order ID") AS Order_Count FROM orders
GROUP BY Market;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT MAX(Sales) AS Highest_Sales, MIN(Sales) AS Lowest_Sales FROM orders;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT Market, SUM(Sales) AS TOTAL_SALES FROM orders
GROUP BY Market
HAVING SUM(Sales) > 10000000
ORDER BY TOTAL_SALES DESC;'''
pd.read_sql_query(query, conn)
```

```python
query = '''SELECT Market, COUNT(DISTINCT "Order ID") AS UNIQUE_ORDERS FROM orders
GROUP BY Market;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales, 
CASE
WHEN Sales >= 1000 THEN 'HIGH'
ELSE 'LOW'
END AS Sales_Category
FROM orders
LIMIT 10;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales, Market FROM orders
WHERE Sales > 1000
AND Market = 'Europe'
LIMIT 10;'''
pd.read_sql_query(query,conn)
```

```python
query = '''SELECT "Order ID", Sales, Market FROM orders
WHERE Market IN ('Europe', 'LATAM')
LIMIT 10;'''
pd.read_sql_query(query, conn)
```

```python
query = '''SELECT "Order ID", Market FROM orders
WHERE Market LIKE 'E%'
LIMIT 10;'''
pd.read_sql_query(query,conn)
```
